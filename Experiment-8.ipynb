```python
import torch
from diffusers import AutoPipelineForText2Image

# Load the pre-trained image generation model
model_id = "stabilityai/sd-turbo"

print("Loading image generation model...")

pipe = AutoPipelineForText2Image.from_pretrained(
    model_id,
    torch_dtype=torch.float32
)

# Use CPU
pipe = pipe.to("cpu")

# Text prompt
prompt = "A futuristic smart city with flying cars, green buildings, and artificial intelligence technology"

# Generate image
image = pipe(
    prompt=prompt,
    num_inference_steps=1,
    guidance_scale=0.0
).images[0]

# Display image
display(image)
```

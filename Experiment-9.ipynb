```python
from transformers import AutoImageProcessor, AutoModelForImageClassification
from PIL import Image
import torch

# ==========================================
# EXPERIMENT 9: MULTIMODAL AI
# ==========================================

# Lightweight image classification model
model_name = "google/mobilenet_v2_1.0_224"

print("Loading model...")

# Load processor and model
processor = AutoImageProcessor.from_pretrained(model_name)

model = AutoModelForImageClassification.from_pretrained(
    model_name
)

# Load input image
image = Image.open("image.jpg").convert("RGB")

# Process image
inputs = processor(
    images=image,
    return_tensors="pt"
)

# Make prediction
with torch.no_grad():
    outputs = model(**inputs)

# Get predicted class
predicted_index = torch.argmax(
    outputs.logits,
    dim=1
).item()

predicted_label = model.config.id2label[predicted_index]

# Display results
print("\n========== MULTIMODAL AI RESULT ==========")

print("\nImage Classification:")
print("Predicted Object:", predicted_label)

# Generate text description
description = (
    "The AI analyzed the input image and identified "
    "the main object as: " + predicted_label
)

print("\nGenerated Text Description:")
print(description)
```

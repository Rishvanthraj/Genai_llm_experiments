```python
# ==========================================
# EXPERIMENT 11
# AI-BASED MULTIMEDIA CONTENT GENERATION
# ==========================================

from gtts import gTTS
from PIL import Image, ImageDraw
import textwrap

# ------------------------------------------
# 1. HIGH-LEVEL USER INPUT
# ------------------------------------------

topic = "Artificial Intelligence in Education"

# ------------------------------------------
# 2. GENERATE TEXT CONTENT
# ------------------------------------------

generated_text = f"""
{topic}

Artificial Intelligence is transforming education by making learning more
personalized, interactive, and accessible. AI-powered systems can help students
understand difficult concepts, provide instant feedback, and recommend learning
materials based on individual performance.

Teachers can also use AI to automate repetitive tasks and focus more on
teaching and student development. Therefore, Artificial Intelligence has the
potential to improve the overall learning experience.
"""

print("========== GENERATED TEXT ==========\n")
print(generated_text)

# ------------------------------------------
# 3. CREATE A SIMPLE IMAGE
# ------------------------------------------

image = Image.new("RGB", (800, 500), "white")

draw = ImageDraw.Draw(image)

image_text = "AI IN EDUCATION\nSmart Learning for the Future"

draw.multiline_text(
    (150, 200),
    image_text,
    fill="black",
    spacing=20,
    align="center"
)

image.save("generated_image.png")

print("\nImage generated successfully!")
display(image)

# ------------------------------------------
# 4. TEXT-TO-SPEECH AUDIO GENERATION
# ------------------------------------------

tts = gTTS(
    text=generated_text,
    lang="en"
)

tts.save("generated_audio.mp3")

print("\nAudio generated successfully!")
print("File saved as: generated_audio.mp3")

# ------------------------------------------
# FINAL OUTPUT
# ------------------------------------------

print("\n========== MULTIMEDIA CONTENT GENERATED ==========")
print("1. Text Content")
print("2. Image: generated_image.png")
print("3. Audio: generated_audio.mp3")
```
